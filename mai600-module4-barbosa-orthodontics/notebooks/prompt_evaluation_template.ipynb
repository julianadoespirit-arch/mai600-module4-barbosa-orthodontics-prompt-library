# MAI 600 Module 4 – Prompt Evaluation Template

**Project:** AI Prompt Library for Orthodontic Patient Consultation Follow-Up  
**Case Study:** Barbosa Orthodontics

This notebook loads the fictional evaluation cases, stores the five prompts, and provides a structure for testing and scoring model outputs.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('../evaluation_set.csv')
cases = pd.read_csv(DATA_PATH)
cases.head()
```

```python
PROMPTS = {
    'P1': '''Summarize the orthodontic consultation in five bullets: patient concern, treatment interest, financial or insurance question, scheduling preference, and recommended administrative next action. Use only the information provided.''',
    'P2': '''You are an administrative assistant supporting the treatment coordinator at Barbosa Orthodontics. Analyze the fictional consultation note using only the information provided. Return a markdown table with columns: Patient Concern, Treatment Interest, Financial/Insurance Questions, Scheduling Preferences, Recommended Administrative Next Action, and Missing Information. If unavailable, write Not provided. Do not provide medical advice.''',
    'P3': '''Use the few-shot prompt stored in prompt_library.md.''',
    'P4': '''Analyze only the provided note. Do not diagnose, provide medical advice, promise insurance coverage, invent financing, or invent appointment details. Use Not provided for missing information. Recommend administrative actions only.''',
    'P5': '''You are the AI Patient Follow-Up Assistant for Barbosa Orthodontics. Summarize the fictional note in exactly one markdown table with columns: Patient Concern or Goal, Treatment Interest, Financial or Insurance Questions, Scheduling Preferences, Recommended Administrative Next Action, Missing Information. Use only the note. Do not provide medical advice or invent treatment, insurance, financing, scheduling, or patient details. Write Not provided when information is missing. Use a concise, respectful, professional tone.'''
}

list(PROMPTS.keys())
```

```python
def build_full_prompt(prompt_id: str, consultation_note: str) -> str:
    if prompt_id not in PROMPTS:
        raise ValueError(f'Unknown prompt ID: {prompt_id}')
    return f"{PROMPTS[prompt_id]}\n\nConsultation note:\n{consultation_note}"

# Example
print(build_full_prompt('P5', cases.loc[0, 'Input / Scenario']))
```

```python
score_columns = [
    'Accuracy', 'Helpfulness', 'Format Adherence', 'Completeness',
    'Evidence/Grounding', 'Safety/Professionalism', 'Clarity'
]

evaluation_rows = []

for _, row in cases.iterrows():
    evaluation_rows.append({
        'Test ID': row['Test ID'],
        'Prompt Used': 'P5',
        'Model Output': '',
        **{col: None for col in score_columns},
        'Failure Mode': '',
        'Evaluator Notes': ''
    })

evaluation_df = pd.DataFrame(evaluation_rows)
evaluation_df.head()
```

```python
# After adding model outputs and scores, save the completed evaluation.
OUTPUT_PATH = Path('../notebook_results.csv')
evaluation_df.to_csv(OUTPUT_PATH, index=False)
print(f'Saved template to: {OUTPUT_PATH.resolve()}')
```
